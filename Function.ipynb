```python
#password strength checker
def is_strong_password(password):
    if  len(password)<8:
        return False
    if not any(char.isdigit() for char in password):
         return False
    if not any(char.islower() for char in password):
        return False
    if not any(char.isupper() for char in password):
        return False
    return True
print(is_strong_password("hdhfb"))  
```

```python
def calculate_total_cost(cart):
    total_cost = 0
    for item in cart:
        total_cost += item['price'] * item['quantity']
    
    return total_cost

cart = [
    {'name': 'apple', 'price': 30, 'quantity': 3},
    {'name': 'banana', 'price': 45, 'quantity': 6}
]

print(calculate_total_cost(cart))
```

```python
def linear_search(arr,taraget):
    size=len(arr)
    for index in range(0,size):
        if(arr[index]==taraget):
            return index
    return -1
my_list=[1,2,3,4,5,6]
taraget=3
result=linear_search(my_list,taraget)
print(result)
  
```
